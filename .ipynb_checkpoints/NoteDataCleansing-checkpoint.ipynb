```python
from multiprocessing.reduction import duplicate

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler

train = pd.read_csv(
    r'/HousingDataLearning\house-prices-advanced-regression-techniques\train.csv')

print(train.info())  # info
print(train.describe(include='all'))  # analisis deskriptif

missing_values = train.isnull().sum()
missing_values = missing_values[missing_values > 0]  # memeriksa jumlah nilai yang kosong di tiap kolom

# eliminasi data hilang (diatas 75%)
less = missing_values[missing_values < 1000].index
over = missing_values[missing_values >= 1000].index


# eliminasi data untuk numerik
numeric_features = train[less].select_dtypes(
    include='number').columns  # memilh data numerik dari kolom yang ditentukan less
train[numeric_features] = train[numeric_features].fillna(
    train[numeric_features].median())  # semua nilai kolom hilang NaN diisi median

df = train.drop(columns=over)  # hapus kolom dengan data hilang lebih dari 75%

numeric_features = df.select_dtypes(include='number').columns

# eliminasi data untuk object/str
kategorical_features = train[less].select_dtypes(include='string').columns

for column in kategorical_features:
    train[column] = train[column].fillna(train[column].mode()[0])

# cek ulang
missing_values = df.isnull().sum()
missing_values = missing_values[missing_values > 0]

#buat munculin sebelum outliers
"""
for feature in numeric_features:
    plt.figure(figsize = (10,6))
    sns.boxplot(x=df[feature])
    plt.title(f'Box Plot of {feature}')
    plt.show()
"""
# Remove outlier Q1 dan Q3
Q1 = df[numeric_features].quantile(0.25)
Q3 = df[numeric_features].quantile(0.75)
IQR = Q3 - Q1

# Filter data data yang tidak mengandung outliers
condition = ~(((df[numeric_features]) < (Q1 - 1.5 * IQR)) |
              ((df[numeric_features]) > (Q3 + 1.5 * IQR))).any(axis=1)

df_filtered_numeric = df.loc[condition, numeric_features]

# Menggabungkan dengan kolom kategorikal kembali
categorical_features = df.select_dtypes(include='string').columns
df = pd.concat([df_filtered_numeric, df.loc[condition, categorical_features]], axis=1)

"""
for feature in numeric_features:
    plt.figure(figsize = (10,6))
    sns.boxplot(x=df[feature])
    plt.title(f'Box Plot of {feature}')
    plt.show()
"""

# standarisasi fitur numerik
scaler = StandardScaler()
df[numeric_features] = scaler.fit_transform(df[numeric_features])

category_features = df.select_dtypes(include='object').columns
df[category_features]

#mengecek baris duplikat
"""
duplicates = df.duplicated()
print("baris duplikat :")
print(df[duplicates])
"""
# Histogram Sebelum Standardisasi
"""
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
sns.histplot(train[numeric_features[3]], kde=True)
plt.title("Histogram Sebelum Standardisasi")

plt.subplot(1, 2, 2)
sns.histplot(df[numeric_features[3]], kde=True)
plt.title("Histogram Sesudah Standardisasi")

plt.tight_layout
plt.show()
"""

#buat ngecek
"""
print(numeric_features)
print(len(numeric_features))
"""
```
